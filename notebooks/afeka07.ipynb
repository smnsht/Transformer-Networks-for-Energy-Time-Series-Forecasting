```python
import pandas as pd
```

```python
df_15min = pd.read_csv('../data/afeka/germany_thin.csv', parse_dates=['utc_timestamp'])
df_15min
```

```python
data = df_15min.copy()

# Add percentage error
data['perr'] = (data['load_actual_entsoe_transparency'] - data['load_forecast_entsoe_transparency']) / data['load_actual_entsoe_transparency']
data
```

# Predictions from ENTSOE have MAPE of 3%, while best predictions obtained by models from this project are around 2.5%!!!

```python
# Compute Mean Absolute Percentage Error (MAPE)
mape = data['perr'].abs().mean()

print("Mean Absolute Percentage Error (MAPE):", mape)
```
